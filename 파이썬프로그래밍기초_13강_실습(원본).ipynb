**데이터 전처리(데이터 변환 및 불용어 필터링)**

```python
# 특수문자 목록
PUNCTUATIONS = ".,!?\"':;()-"

def clean_text(fp):
    text = ""

    for line in fp:
        #소문자 변환
        line = line.lower()

        for char in PUNCTUATIONS:
            line = line.replace(char, "")
        text += " " + line
    return text


```
